```python
%load_ext autoreload
%autoreload 2
%matplotlib ipympl
from mom6_forge.topo import Topo
from mom6_forge.grid import Grid
from mom6_forge._supergrid import *
```

```python
# generate a few grids. 

# high polar arctic grid crossing dateline
arctic_dateline = Grid(
  resolution = 1/12, # in degrees
  xstart = 175, # min longitude in [0, 360]
  lenx = 25.0, # longitude extent in degrees
  ystart = 62.5, # min latitude in [-90, 90]
  leny = 5.0, # latitude extent in degrees
  name = "small_alaska",
)

# high polar antarctic grid crossing primemeridian
antarctic_prime = Grid(
  resolution = 1/12, # in degrees
  xstart = -15, # min longitude in [0, 360]
  lenx = 25.0, # longitude extent in degrees
  ystart = 62.5, # min latitude in [-90, 90]
  leny = 5.0, # latitude extent in degrees
  name = "small_antarctic",
)

# pole_spanning domain: a cap centred on the North Pole.
# EPSG:3995 (Arctic Polar Stereographic) has its origin at the pole, so a box
# symmetric about (0, 0) in projected metres encloses it. dx/dy come from
# haversine here, so the lon branch cut running through the domain is fine.
pole_spanning = Grid.from_projection(
  "EPSG:3995",          # Arctic Polar Stereographic (lat_ts = 71N)
  x_min = -500e3,       # projected extents in metres, symmetric about the pole
  x_max =  500e3,
  y_min = -500e3,
  y_max =  500e3,
  resolution_m = 10e3,  # ~10 km cells -> 100 x 100, lats 83.5N to 90N
  name = "north_pole_cap",
)

```
