```python
from pathlib import Path
from mom6_forge.mapping import *
rof_esmf_mesh_filepath = "/glade/campaign/cesm/cesmdata/cseg/inputdata/ocn/mom/croc/rof/glofas/dis24/GLOFAS_esmf_mesh_v4.nc"
ocn_mesh_filepath = "/glade/work/ajanney/MOM6_MPAS_CARIB/Variable_Res_IndoPacific/rmom6_input/CrocIndoPacific_112/ocnice/ESMF_mesh_CrocIndoPacific_CrocIndoPacific_112_45f297.nc"
# NOTE: this must be a real directory, not "" - an empty output_dir resolves
# to the notebook's cwd, which would dump multi-GB mapping files straight
# into this git-tracked dev/ folder.
mapping_dir = Path("~/scratch/croc_input/runoff_mapping_scaling/basic_rof_mapping_call_test").expanduser()
mapping_dir.mkdir(parents=True, exist_ok=True)
mapping_file_prefix = "new_prefix"
rmax = 20
fold = 20
```
